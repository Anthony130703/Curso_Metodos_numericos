**A) Demostración de la ecuación para la longitud de onda máxima**

Partimos de la ley de radiación de Planck, que describe la intensidad de radiación por unidad de área y por unidad de longitud de onda para un cuerpo negro a una temperatura $T$:
$$I(\lambda) = \frac{2\pi h c^2 \lambda^{-5}}{e^{\frac{hc}{\lambda k_B T}} - 1}$$

Para simplificar la derivación analítica, agrupamos las constantes físicas en $C = 2\pi h c^2$ y definimos $a = \frac{hc}{k_B T}$. La función se reescribe como:
$$I(\lambda) = C \lambda^{-5} (e^{a/\lambda} - 1)^{-1}$$

Para hallar la longitud de onda donde la intensidad es máxima, derivamos $I(\lambda)$ con respecto a $\lambda$ e igualamos a cero, aplicando la regla del producto y de la cadena:
$$\frac{dI}{d\lambda} = C \left[ -5\lambda^{-6} (e^{a/\lambda} - 1)^{-1} + \lambda^{-5} (-1)(e^{a/\lambda} - 1)^{-2} \cdot e^{a/\lambda} \cdot \left(-\frac{a}{\lambda^2}\right) \right] = 0$$

Factorizamos los términos comunes $C \lambda^{-6} (e^{a/\lambda} - 1)^{-1}$ para ordenar la expresión:
$$C \lambda^{-6} (e^{a/\lambda} - 1)^{-1} \left[ -5 + \frac{a}{\lambda} \left( \frac{e^{a/\lambda}}{e^{a/\lambda} - 1} \right) \right] = 0$$

Dado que las constantes y el primer factor no pueden ser cero para una longitud de onda térmica finita, el interior del corchete debe ser obligatoriamente cero:
$$-5 + \frac{a}{\lambda} \left( \frac{e^{a/\lambda}}{e^{a/\lambda} - 1} \right) = 0$$

Aplicamos el cambio de variable sugerido por el problema $x = \frac{hc}{\lambda k_B T}$, lo cual equivale a sustituir $x = \frac{a}{\lambda}$:
$$-5 + x \left( \frac{e^x}{e^x - 1} \right) = 0$$

Multiplicamos toda la ecuación por el denominador $(e^x - 1)$ para eliminar la fracción:
$$-5(e^x - 1) + x e^x = 0$$
$$-5e^x + 5 + x e^x = 0$$

Finalmente, dividimos todos los términos entre $e^x$ (lo que equivale a multiplicar por $e^{-x}$) para obtener la ecuación trascendente final:
$$5e^{-x} + x - 5 = 0$$
