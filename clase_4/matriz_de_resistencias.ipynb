## Derivación del Sistema de Ecuaciones para la Red de Resistores

Para encontrar la ecuación que verifican los potenciales en cada nodo, aplicaremos la **Ley de Corrientes de Kirchhoff**. Esta ley establece que la suma algebraica de las corrientes que salen de cualquier nodo debe ser igual a cero ($\sum I_{salida} = 0$).

Por la Ley de Ohm, la corriente que fluye desde un nodo $V_a$ hacia un nodo $V_b$ a través de una resistencia $R$ es:
$$ I = \frac{V_a - V_b}{R} $$

Dado que en este arreglo **todas las resistencias tienen el mismo valor $R$**, el factor $\frac{1}{R}$ será común en todos los términos y se podrá simplificar (multiplicando toda la ecuación por $R$), dejando únicamente la relación entre los voltajes.

A continuación, analizamos cada nodo del diagrama:

### 1. Análisis del Nodo 1 ($V_1$)
El nodo 1 está conectado a la fuente $V_+$, al nodo $V_2$ y al nodo $V_3$.
$$ \frac{V_1 - V_+}{R} + \frac{V_1 - V_2}{R} + \frac{V_1 - V_3}{R} = 0 $$
Multiplicando por $R$ y agrupando términos:
$$ 3V_1 - V_+ - V_2 - V_3 = 0 $$
$$ \mathbf{3V_1 - V_2 - V_3 = V_+} \quad \text{--- (Ecuación 1)} $$

### 2. Análisis del Nodo 2 ($V_2$)
El nodo 2 está conectado a la fuente $V_+$, y a los nodos $V_1$, $V_3$ y $V_4$.
$$ \frac{V_2 - V_+}{R} + \frac{V_2 - V_1}{R} + \frac{V_2 - V_3}{R} + \frac{V_2 - V_4}{R} = 0 $$
Multiplicando por $R$ y agrupando términos:
$$ 4V_2 - V_+ - V_1 - V_3 - V_4 = 0 $$
$$ \mathbf{-V_1 + 4V_2 - V_3 - V_4 = V_+} \quad \text{--- (Ecuación 2)} $$

### 3. Nodos Intermedios ($V_i$ para $3 \le i \le N-2$)
Cualquier nodo intermedio se conecta en forma de "cruz" con sus dos nodos numéricamente anteriores ($i-2$, $i-1$) y sus dos nodos numéricamente posteriores ($i+1$, $i+2$).
$$ \frac{V_i - V_{i-2}}{R} + \frac{V_i - V_{i-1}}{R} + \frac{V_i - V_{i+1}}{R} + \frac{V_i - V_{i+2}}{R} = 0 $$
$$ \mathbf{-V_{i-2} - V_{i-1} + 4V_i - V_{i+1} - V_{i+2} = 0} \quad \text{--- (Ecuación } i \text{)} $$

### 4. Análisis del Penúltimo Nodo ($V_{N-1}$)
El nodo $N-1$ se conecta a sus nodos anteriores ($V_{N-3}$, $V_{N-2}$), al nodo final $V_N$ y, crucialmente, **al nodo de tierra (0 Volts)**.
$$ \frac{V_{N-1} - V_{N-3}}{R} + \frac{V_{N-1} - V_{N-2}}{R} + \frac{V_{N-1} - V_N}{R} + \frac{V_{N-1} - 0}{R} = 0 $$
$$ \mathbf{-V_{N-3} - V_{N-2} + 4V_{N-1} - V_N = 0} \quad \text{--- (Ecuación } N-1 \text{)} $$

### 5. Análisis del Último Nodo ($V_N$)
El nodo $N$ se conecta a los nodos $V_{N-2}$, $V_{N-1}$ y también **al nodo de tierra (0 Volts)**.
$$ \frac{V_N - V_{N-2}}{R} + \frac{V_N - V_{N-1}}{R} + \frac{V_N - 0}{R} = 0 $$
$$ \mathbf{-V_{N-2} - V_{N-1} + 3V_N = 0} \quad \text{--- (Ecuación } N \text{)} $$

## Solución Estacionaria del Sistema Masa-Resorte

Para encontrar las **soluciones estacionarias** del sistema físico, debemos considerar que el sistema ha alcanzado el equilibrio. En este estado, las masas ya no se mueven, por lo que las aceleraciones son nulas:
$$ \frac{d^2x_1}{dt^2} = 0, \quad \frac{d^2x_2}{dt^2} = 0, \quad \frac{d^2x_3}{dt^2} = 0 $$

Sustituyendo estas condiciones en las ecuaciones diferenciales originales de la segunda ley de Newton, obtenemos el sistema de ecuaciones algebraicas:

**Masa 1:**
$$ 0 = 2k(x_2 - x_1) + m_1g - kx_1 $$
Agrupando los términos de posición a la izquierda:
$$ 3kx_1 - 2kx_2 = m_1g \quad \text{--- (Ecuación 1)} $$

**Masa 2:**
$$ 0 = k(x_3 - x_2) + m_2g - 2k(x_2 - x_1) $$
Agrupando los términos:
$$ -2kx_1 + 3kx_2 - kx_3 = m_2g \quad \text{--- (Ecuación 2)} $$

**Masa 3:**
$$ 0 = m_3g - k(x_3 - x_2) $$
Agrupando los términos:
$$ -kx_2 + kx_3 = m_3g \quad \text{--- (Ecuación 3)} $$

### Representación Matricial $Ax = b$

Podemos escribir este sistema en forma matricial agrupando los coeficientes de rigidez $k$ en una matriz tridiagonal simétrica:

$$

\end{pmatrix}
\begin{pmatrix}

\end{pmatrix}
=

$$

## El Problema Físico: Densidad de una Varilla (Problema Inverso)

**¿De qué trata este problema?** 
Físicamente, este es un "problema inverso". En la vida real, si conoces de qué está hecha una varilla (su densidad interna $\rho(x)$), puedes usar integrales para calcular su masa total o su centro de masa. Aquí hacemos lo contrario: **solo podemos medir el exterior** (la masa $m$ en una balanza y el centro de masa $x_{cm}$ equilibrando la varilla en un filo), y queremos usar esos datos para "adivinar" cómo está distribuida la masa en su interior. 

Para lograrlo, asumimos que la densidad interna es un polinomio: $\rho(x) = c_1 + c_2x + \dots + c_nx^{n-1}$. Nuestro objetivo es encontrar esos coeficientes $c$.

---

### Parte (a): Construcción de la Matriz de Hilbert

Se nos define que los momentos medidos (masa, centro de masa, inercia) siguen la fórmula integral:
$$ M_k = \int_0^1 x^k \rho(x) dx \quad \text{para } k = 0, 1, \dots, n-1 $$

Sustituimos nuestro polinomio $\rho(x)$ dentro de la integral:
$$ M_k = \int_0^1 x^k (c_1 + c_2x + c_3x^2 + \dots + c_nx^{n-1}) dx $$

Multiplicamos $x^k$ por cada término:
$$ M_k = \int_0^1 (c_1 x^k + c_2 x^{k+1} + c_3 x^{k+2} + \dots + c_n x^{k+n-1}) dx $$

Resolvemos la integral término por término sabiendo que $\int_0^1 x^p dx = \frac{1}{p+1}$:
$$ M_k = c_1 \left(\frac{1}{k+1}\right) + c_2 \left(\frac{1}{k+2}\right) + \dots + c_n \left(\frac{1}{k+n}\right) $$

Esta es una ecuación lineal. Si definimos que la fila de la matriz es $i$ (donde $i = k+1$ para que empiece en 1) y la columna de la matriz es $j$, el factor que multiplica a cada coeficiente $c_j$ es exactamente $h_{ij} = \frac{1}{i+j-1}$. Esta estructura forma la famosa **Matriz de Hilbert ($H_n$)**.

Para $n=2$ (polinomio lineal) y $n=3$ (polinomio cuadrático), las matrices son:
$$ H_2 = \begin{pmatrix} 1 & 1/2 \\ 1/2 & 1/3 \end{pmatrix}, \quad H_3 = \begin{pmatrix} 1 & 1/2 & 1/3 \\ 1/2 & 1/3 & 1/4 \\ 1/3 & 1/4 & 1/5 \end{pmatrix} $$

---

### Parte (b): Densidad Lineal ($n=2$)

**1. Plantear el sistema original:**
*   Datos físicos: Masa $m = 2.5$ kg, Centro de masa $x_{cm} = 0.6$ m.
*   Momentos: 
    *   $M_0 = m = 2.5$.
    *   $M_1 = m \cdot x_{cm} = (2.5)(0.6) = 1.5$.
*   Vector de momentos $M = (2.5, 1.5)^T$.

El sistema $H_2 c = M$ es:
$$ \begin{pmatrix} 1 & 1/2 \\ 1/2 & 1/3 \end{pmatrix} \begin{pmatrix} c_1 \\ c_2 \end{pmatrix} = \begin{pmatrix} 2.5 \\ 1.5 \end{pmatrix} $$

Resolviendo manualmente por Gauss o sustitución:
*   $c_1 + 0.5c_2 = 2.5$
*   $0.5c_1 + 0.333c_2 = 1.5$
*   Resultado: $c_1 = 1$, $c_2 = 3$. La densidad real es $\mathbf{\rho(x) = 1 + 3x}$.

**2. Análisis de Errores (Normas e Inversa a mano):**
Para hallar $H_2^{-1}$, usamos la fórmula de la inversa de una matriz $2 \times 2$ ($\frac{1}{det} \begin{pmatrix} d & -b \\ -c & a \end{pmatrix}$):
*   Determinante de $H_2 = (1)(1/3) - (1/2)(1/2) = 1/3 - 1/4 = 1/12$.
*   $H_2^{-1} = 12 \begin{pmatrix} 1/3 & -1/2 \\ -1/2 & 1 \end{pmatrix} = \mathbf{\begin{pmatrix} 4 & -6 \\ -6 & 12 \end{pmatrix}}$.

Calculamos las Normas de Frobenius ($||A||_F = \sqrt{\sum a_{ij}^2}$):
*   $||H_2||_F = \sqrt{1^2 + 0.5^2 + 0.5^2 + (1/3)^2} = \sqrt{1.611} \approx \mathbf{1.269}$
*   $||H_2^{-1}||_F = \sqrt{4^2 + (-6)^2 + (-6)^2 + 12^2} = \sqrt{232} \approx \mathbf{15.231}$
*   Número de condición $\kappa_F(H_2) = 1.269 \times 15.231 \approx \mathbf{19.33}$.

**3. Perturbación (Error de medición):**
Si medimos $x_{cm}$ con un milímetro de error ($x_{cm} = 0.601$ m):
*   Nuevo $M_1 = 2.5 \times 0.601 = 1.5025$. 
*   Nuevo vector $M_{err} = (2.5, 1.5025)^T$.
*   Vector de error en datos $\delta M = (0, 0.0025)^T$.
*   Calculamos los nuevos coeficientes multiplicando por la inversa: $c_{err} = H_2^{-1} M_{err} = \begin{pmatrix} 0.985 \\ 3.03 \end{pmatrix}$.
*   Vector de error en la solución $\delta c = c_{err} - c = (-0.015, 0.03)^T$.

**Verificación de la cota matemática** ($\frac{||\delta c||_2}{||c||_2} \le \kappa_F \frac{||\delta M||_2}{||M||_2}$):
*   Error relativo en M: $0.0025 / \sqrt{2.5^2 + 1.5^2} \approx 0.00085$ (0.085%)
*   Error relativo en c: $\sqrt{(-0.015)^2 + 0.03^2} / \sqrt{1^2 + 3^2} \approx 0.0106$ (1.06%)
*   Cota: $0.0106 \le 19.33 \times 0.00085 \implies 0.0106 \le 0.0164$. ¡Se cumple la teoría! Un error diminuto en la regla amplificó el error de la densidad más de 12 veces.

---

### Parte (c): Densidad Cuadrática ($n=3$)
Aquí la varilla verdadera tiene $\rho(x) = 3 + 3x + 3x^2$, y medimos los momentos $M = (5.5, 3.25, 2.35)^T$.
La matriz de Hilbert $H_3$ es extremadamente inestable. Su inversa exacta es:
$$ H_3^{-1} = \begin{pmatrix} 9 & -36 & 30 \\ -36 & 192 & -180 \\ 30 & -180 & 180 \end{pmatrix} $$

*   Si calculamos $\kappa_F(H_3) \approx 450$. El número de condición creció violentamente desde $19$ hasta $450$. Esto significa que si agregamos una tercera variable (inercia $M_2$) al modelo, el sistema se vuelve matemáticamente caótico frente a pequeños errores de medición.
*   Si perturbamos $M_2$ con solo un 1% de error ($M_2 = 2.3735$), la multiplicación $H_3^{-1} M_{err}$ arroja coeficientes completamente destruidos (alejados de 3, 3, 3) porque el $\kappa_F=450$ multiplicará el error del 1% convirtiéndolo en casi un 450% de error en la curva final.

## Sistema Masa-Resorte: Equilibrio Estático y Condicionamiento

### (a) Planteamiento del Equilibrio Estático ($Kx = F$)

Para encontrar la ecuación de la matriz de rigidez, aplicamos la Primera Ley de Newton ($\sum F = 0$) a cada masa cuando el sistema está quieto (equilibrio). Usamos la Ley de Hooke, donde la fuerza de un resorte es proporcional a cuánto se estira o comprime ($F_k = k \Delta x$).

**Análisis de la Masa 1 ($m_1$):**
Sobre la masa 1 actúan tres fuerzas:
1.  El resorte 1 la jala hacia la pared (izquierda) si se mueve a la derecha: $-k_1 x_1$
2.  El resorte 2 la jala hacia la derecha si $m_2$ se mueve más que $m_1$: $+k_2 (x_2 - x_1)$
3.  La fuerza externa aplicada directamente sobre ella: $+F_1$

En equilibrio, la suma de estas fuerzas es cero:
$$ -k_1 x_1 + k_2 (x_2 - x_1) + F_1 = 0 $$
Despejamos $F_1$ y factorizamos:
$$ (k_1 + k_2)x_1 - k_2 x_2 = F_1 $$

**Análisis de la Masa 2 ($m_2$):**
Sobre la masa 2 actúan solo dos fuerzas:
1.  El resorte 2 la jala hacia la izquierda si ella se aleja de $m_1$: $-k_2 (x_2 - x_1)$
2.  La fuerza externa aplicada directamente sobre ella: $+F_2$

En equilibrio:
$$ -k_2 (x_2 - x_1) + F_2 = 0 $$
Despejamos $F_2$:
$$ -k_2 x_1 + k_2 x_2 = F_2 $$

**Forma Matricial:**
Agrupando ambas ecuaciones, formamos el sistema $Kx = F$:
$$ \begin{pmatrix} k_1 + k_2 & -k_2 \\ -k_2 & k_2 \end{pmatrix} \begin{pmatrix} x_1 \\ x_2 \end{pmatrix} = \begin{pmatrix} F_1 \\ F_2 \end{pmatrix} $$

Sustituyendo los valores dados ($k_1 = 1$, $k_2 = 1000$):
$$ K = \begin{pmatrix} 1001 & -1000 \\ -1000 & 1000 \end{pmatrix} $$

---

### (b) Cálculo de la Inversa y Número de Condición

Para hallar la inversa $K^{-1}$ de una matriz $2 \times 2$, usamos la fórmula $\frac{1}{det(K)} \begin{pmatrix} d & -b \\ -c & a \end{pmatrix}$.
1.  **Determinante:** $det(K) = (1001)(1000) - (-1000)(-1000) = 1001000 - 1000000 = 1000$.
2.  **Inversa:** 
$$ K^{-1} = \frac{1}{1000} \begin{pmatrix} 1000 & 1000 \\ 1000 & 1001 \end{pmatrix} = \begin{pmatrix} 1 & 1 \\ 1 & 1.001 \end{pmatrix} $$

**Cálculo de Normas de Frobenius ($||A||_F = \sqrt{\sum a_{ij}^2}$):**
*   $||K||_F = \sqrt{1001^2 + (-1000)^2 + (-1000)^2 + 1000^2} = \sqrt{4002001} \approx 2000.5$
*   $||K^{-1}||_F = \sqrt{1^2 + 1^2 + 1^2 + 1.001^2} = \sqrt{4.002001} \approx 2.0005$

**Número de Condición:**
$$ \kappa_F(K) = ||K||_F \cdot ||K^{-1}||_F = 2000.5 \times 2.0005 \approx 4002 $$

*Pérdida de Cifras:* Se pierden aproximadamente $\log_{10}(\kappa_F)$ cifras significativas. En este caso, $\log_{10}(4002) \approx 3.6$. **Se pierden entre 3 y 4 cifras significativas** al resolver este sistema.

---

### (c) Fuerzas Iguales y Opuestas

Se aplican $F_1 = -1$ N y $F_2 = 1$ N.
Para hallar los desplazamientos, multiplicamos la inversa por el vector de fuerzas:
$$ x = K^{-1} F = \begin{pmatrix} 1 & 1 \\ 1 & 1.001 \end{pmatrix} \begin{pmatrix} -1 \\ 1 \end{pmatrix} = \begin{pmatrix} -1 + 1 \\ -1 + 1.001 \end{pmatrix} = \begin{pmatrix} 0 \\ 0.001 \end{pmatrix} \text{ m} $$

**Interpretación Física:**
El resorte rígido $k_2$ (que está en el medio) recibe toda la tensión y se estira $1$ milímetro ($x_2 - x_1 = 0.001$ m). La masa 1 no se mueve ($x_1 = 0$) porque la fuerza externa que la empuja hacia la pared ($-1$ N) se cancela perfectamente con la tensión del resorte rígido que la jala hacia la derecha ($+1$ N). El resorte blando $k_1$ ni se entera de lo que está pasando.

---

### (d) El Error de Medición y la Inestabilidad

Por un error, medimos $F_1 = -0.999$ N. La fuerza neta ha cambiado.
El vector de error es $\delta F = (0.001, 0)^T$ N.

Resolvemos el sistema perturbado:
$$ x_{nuevo} = K^{-1} \begin{pmatrix} -0.999 \\ 1 \end{pmatrix} = \begin{pmatrix} 1 & 1 \\ 1 & 1.001 \end{pmatrix} \begin{pmatrix} -0.999 \\ 1 \end{pmatrix} = \begin{pmatrix} 0.001 \\ 0.002 \end{pmatrix} \text{ m} $$
El error en el desplazamiento es $\delta x = x_{nuevo} - x = (0.001, 0.001)^T$ m.

**Errores Relativos:**
*   Error en la fuerza: $\frac{||\delta F||_2}{||F||_2} = \frac{0.001}{\sqrt{(-1)^2 + 1^2}} = \frac{0.001}{1.414} \approx 0.0007$ (Un minúsculo 0.07%)
*   Error en la posición: $\frac{||\delta x||_2}{||x||_2} = \frac{\sqrt{0.001^2 + 0.001^2}}{0.001} = \frac{0.00141}{0.001} \approx 1.41$ (Un masivo 141%)

**Verificación de la Cota Matemática:**
$$ \frac{||\delta x||_2}{||x||_2} \le \kappa_F(K) \frac{||\delta F||_2}{||F||_2} \implies 1.41 \le 4002 \times 0.0007 \implies \mathbf{1.41 \le 2.80} $$
La regla se cumple. El factor de amplificación real fue de aproximadamente $2000$ veces (pasamos de un error del 0.07% a uno del 141%).

**Explicación Física (La respuesta a la última pregunta):**
Ese pequeño error de $0.001$ N introduce una *fuerza neta* hacia la derecha sobre todo el sistema. Como el resorte central ($k_2=1000$) es tan rígido, las dos masas actúan casi como un solo bloque sólido. Por lo tanto, el único encargado de resistir y absorber esa fuerza extra es el resorte blando de la pared ($k_1=1$). Para que un resorte tan débil contrarreste una fuerza de $0.001$ N, necesita estirarse muchísimo: exactamente $0.001$ metros ($1$ mm). Por eso, ambas masas se desplazan $1$ mm hacia la derecha.
