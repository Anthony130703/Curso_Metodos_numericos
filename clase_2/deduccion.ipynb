### Deducción del Resto de Lagrange - Serie de Taylor para sin(x)

**1. Definición del error (Resto) como la diferencia entre la función y el polinomio:**
$$R_n(x) = \sin(x) - P_n(x)$$

**2. Polinomio de Taylor para el seno truncado en el término $n$ (grado $2n+1$):**
$$P_n(x) = x - \frac{x^3}{3!} + \frac{x^5}{5!} - \dots + \frac{(-1)^n x^{2n+1}}{(2n+1)!}$$

**3. Al ser la derivada par igual a 0, evaluamos el resto en la derivada $2n+3$:**
$$R_n(x) = \frac{f^{(2n+3)}(\xi)}{(2n+3)!} x^{2n+3}$$

**4. Tomamos el valor absoluto en ambos lados para acotar el error máximo:**
$$|R_n(x)| = \left| \frac{f^{(2n+3)}(\xi)}{(2n+3)!} x^{2n+3} \right|$$

**5. Como $f(x) = \sin(x)$, sus derivadas siempre están acotadas por 1:**
$$|f^{(2n+3)}(\xi)| \leq 1$$

**6. Sustituyendo este valor máximo, obtenemos la desigualdad final:**
$$|R_n(x)| \leq \frac{|x|^{2n+3}}{(2n+3)!}$$

### Cálculo manual del error de truncamiento para $x = 1.3$ rad

De acuerdo con la definición del resto de Lagrange para la serie truncada del $\sin(x)$, la cota superior del error absoluto está dada por:
$$|R_n(x)| \leq \frac{|x|^{2n+3}}{(2n+3)!}$$

Para garantizar un error de truncamiento menor a $10^{-6}$ evaluando en $x = 1.3$, se establece la siguiente inecuación:
$$\frac{1.3^{2n+3}}{(2n+3)!} < 10^{-6}$$

Se evalúan los valores de $n$ de forma iterativa para encontrar el mínimo entero que cumpla la condición:

* **Para $n = 2$:**
  $$\frac{1.3^{2(2)+3}}{(2(2)+3)!} = \frac{1.3^7}{7!} \approx \frac{6.2748}{5040} \approx 1.24 \times 10^{-3} > 10^{-6}$$

* **Para $n = 3$:**
  $$\frac{1.3^{2(3)+3}}{(2(3)+3)!} = \frac{1.3^9}{9!} \approx \frac{10.6045}{362880} \approx 2.92 \times 10^{-5} > 10^{-6}$$

* **Para $n = 4$:**
  $$\frac{1.3^{2(4)+3}}{(2(4)+3)!} = \frac{1.3^{11}}{11!} \approx \frac{17.9216}{39916800} \approx 4.49 \times 10^{-7} < 10^{-6}$$

**Conclusión:**
El mínimo valor teórico requerido es **$n = 4$**. Esto implica que la sumatoria debe iterar desde $k=0$ hasta $k=4$, calculando un total de 5 términos del polinomio para asegurar la tolerancia exigida.
