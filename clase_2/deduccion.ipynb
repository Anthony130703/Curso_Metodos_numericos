### Deducción del Resto de Lagrange - Serie de Taylor para sin(x)

**1. Definición del error (Resto) como la diferencia entre la función y el polinomio:**
$$R_n(x) = \sin(x) - P_n(x)$$

**2. Polinomio de Taylor para el seno truncado en el término $n$ (grado $2n+1$):**
$$P_n(x) = x - \frac{x^3}{3!} + \frac{x^5}{5!} - \dots + \frac{(-1)^n x^{2n+1}}{(2n+1)!}$$

**3. Al ser la derivada par igual a 0, evaluamos el resto en la derivada $2n+3$:**
$$R_n(x) = \frac{f^{(2n+3)}(\xi)}{(2n+3)!} x^{2n+3}$$

**4. Tomamos el valor absoluto en ambos lados para acotar el error máximo:**
$$|R_n(x)| = \left| \frac{f^{(2n+3)}(\xi)}{(2n+3)!} x^{2n+3} \right|$$

**5. Como $f(x) = \sin(x)$, sus derivadas siempre están acotadas por 1:**
$$|f^{(2n+3)}(\xi)| \leq 1$$

**6. Sustituyendo este valor máximo, obtenemos la desigualdad final:**
$$|R_n(x)| \leq \frac{|x|^{2n+3}}{(2n+3)!}$$
