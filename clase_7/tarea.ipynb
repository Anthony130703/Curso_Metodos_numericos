### 2. Demostración del modelo sin intercepto ($v = H_0d$)
El objetivo es encontrar el parámetro $H_0$ que minimice la suma de los cuadrados de los residuos ($S_r$). La función de error se define como:

$$S_r = \sum_{i=1}^{n} (v_i - H_0 d_i)^2$$

Para encontrar el mínimo, derivamos $S_r$ respecto a $H_0$ e igualamos a cero (aplicando la regla de la cadena):

$$\frac{\partial S_r}{\partial H_0} = \sum_{i=1}^{n} 2(v_i - H_0 d_i)(-d_i) = 0$$

Dividimos entre $-2$ y distribuimos la sumatoria:

$$\sum_{i=1}^{n} (d_i v_i - H_0 d_i^2) = 0$$

$$\sum_{i=1}^{n} d_i v_i - H_0 \sum_{i=1}^{n} d_i^2 = 0$$

Finalmente, despejamos $H_0$ para obtener la fórmula teórica requerida:

$$H_0 = \frac{\sum d_i v_i}{\sum d_i^2}$$

### 3. Comparación de Modelos e Interpretación Física
**Análisis del Error Estándar ($S_{y/x}$) e Intercepto ($a_0$):**
Al comparar ambos ajustes, obtenemos resultados de dispersión muy similares:
* Modelo con intercepto: $S_{y/x} \approx 232.9$
* Modelo sin intercepto: $S_{y/x} \approx 229.0$

El valor absoluto del intercepto en el primer modelo es $|a_0| \approx 40.78$. Observamos que **$|a_0|$ es muy pequeño frente a $S_{y/x}$** ($40.78 \ll 232.9$). Estadísticamente, esto significa que el intercepto se encuentra "ahogado" dentro del margen de ruido e incertidumbre de los datos. En términos prácticos, es indistinguible de cero, lo que justifica utilizar el modelo simplificado $v = H_0d$.

**Significado físico de un intercepto distinto de cero:**
Si el intercepto $a_0$ fuera real y significativamente distinto de cero, implicaría que a una distancia $d = 0$ (es decir, en nuestra propia ubicación espacial), existiría una velocidad de recesión intrínseca. Físicamente, esto significaría que la Vía Láctea tendría un movimiento especial de expansión absoluta respecto al resto del universo, o que el espacio mismo se expande incluso a distancias nulas, lo cual viola el Principio Cosmológico (el cual postula que el universo es isotrópico, homogéneo y sin puntos privilegiados).

### 3. Análisis del Sesgo: ¿Por qué falla la linealización ignorando el fondo $B$?

Al aplicar el logaritmo asumiendo que no hay radiación de fondo, forzamos a los datos a comportarse como una línea recta perfecta descendente: 
$$\ln N = \ln N_0 - \lambda t$$

Sin embargo, en la realidad, para tiempos grandes ($t$ elevado), el material radiactivo se agota y las mediciones del contador Geiger ya no registran la muestra, sino la radiación cósmica ambiental constante (Fondo $B$). Matemáticamente, la curva logarítmica pierde su pendiente constante y se vuelve horizontal.

Al obligar a una regresión lineal clásica a pasar por esa curva aplanada, el modelo calcula una pendiente ($-\lambda$) artificialmente más suave (más cercana a cero). Como el tiempo de vida media es inversamente proporcional a la constante de decaimiento:
$$t_{1/2} = \frac{\ln(2)}{\lambda}$$

Un $\lambda$ subestimado genera una edad artificialmente enorme en nuestro cálculo. Este es un **error sistemático (sesgo)** porque el modelo matemático elegido está equivocado desde su concepción. No importa cuántos datos adicionales tomemos con el contador Geiger; jamás llegaremos al valor real sin usar la regresión no lineal de Gauss-Newton para separar matemáticamente el Fondo $B$ de la muestra radiactiva.
